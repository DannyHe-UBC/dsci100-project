#### Task 4.1: Calculate summary statistics of numeric variables (2 points)

- Use any functions from the `tidyverse` to create **one** table that outputs the mean, minimum, and maximum of **all** numeric columns in your data, dropping the missing values if they exist.

</div>

```r
head(madness_tidy1)
```

```r
names <- madness_tidy1 |>
    select(where(is.numeric), -bid) |>
    names()

means <- madness_tidy1 |>
    select(where(is.numeric), -bid) |>
    map_dbl(mean, na.rm = TRUE)

mins <- madness_tidy1 |>
    select(where(is.numeric), -bid) |>
    map_dbl(min, na.rm = TRUE)

maxes <- madness_tidy1 |>
    select(where(is.numeric), -bid) |>
    map_dbl(max, na.rm = TRUE)

madness_4.1 <- tibble(
    variable = names,
    mean = means,
    minimum = mins,
    maximum = maxes)

madness_4.1
```

Visualization 1: 

This scatter plot shows the relationship between **Regional Win Percentage** and **Total Win Percentage**, with points colored by **bid** type. The strong positive relationship indicates that teams with higher **regional win percentages** generally also achieve higher **total win percentages**. 

The distribution of points suggests that these two variables are informative predictors of **bid** status. The blue points tend to appear more often at the lower and higher ends of the plot, while the red points are more concentrated in the middle range. 

However, the separation between the two bid categories is not very clear, and there is noticeable overlap. This suggests that `reg_wins_pct` and `total_wins_pct` may help distinguish `bid` types, but using only these two variables for prediction may have limited accuracy.



```r
options(repr.plot.width = 10, repr.plot.height = 6)

visual_1 <- ggplot(madness_tidy1,aes(x = reg_wins_pct,y = total_wins_pct)) +
    geom_point(aes(color = bid)) +
    labs(x = "Regional Win Percentage",
        y = "Total Win Percentage",
        color = "Automatic Bid\nor Not",
        title = "Relationship Between Regional\nand Total Win") +
    theme(text = element_text(size = 20))

visual_1
```

The overlap between the two colored distributions helps evaluate whether `total_wins_pct` is a useful predictor of bid status. 

If there is substantial overlap, the variable provides limited ability to distinguish between `bid` categories. If there is little overlap, the variable is more effective for classification. 

In this plot, the distributions partially overlap but also show some separation, suggesting that `total_wins_pc`t can be used to classify `bid` status, although it may not be the most accurate predictor.
